# Predicting Housing Prices with Scikit-Learn

Simple project to learn how to use Scikit-Learn to predict housing prices.


```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
```

```python
# Load the California housing dataset
# as_frame=True returns the data as a pandas DataFrame 
# instead of a MunPy array which is bby default
data = fetch_california_housing(as_frame=True) 
df = pd.DataFrame(data.data, columns=data.feature_names)

df["Price"] = data.target
```
